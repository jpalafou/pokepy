```python
from pokepy import Pokemon, Move, simulate_move
```

```python
tackle          = Move("Tackle", "normal", 40)
leaf_blade      = Move("Leaf Blade", "grass", 90)
flamethrower    = Move("Flamethrower", "fire", 90)
surf            = Move("Surf", "water", 90)

bulbasaur       = Pokemon("Bulbasaur", "grass", 100, 100, 100)
charmander      = Pokemon("Charmander", "fire", 100, 100, 100)
squirtle        = Pokemon("Squirtle", "water", 100, 100, 100)
```

```python
simulate_move(charmander, squirtle, tackle)
```

```python
simulate_move(charmander, squirtle, flamethrower)
```

```python
simulate_move(bulbasaur, squirtle, leaf_blade)
```
